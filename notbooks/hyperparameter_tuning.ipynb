```python
# imports

import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from scipy.stats import randint
from sklearn.model_selection import RandomizedSearchCV
```

```python
# read the data set

df = pd.read_csv('../data/winequalityN.csv')
```

```python
# EDA

df.info()
```

```python
df.isna().sum()
```

```python
df.head()
```

```python
df_copy = df.dropna()
```

```python
df_copy.info()
```

```python
# check type unique values

df_copy['type'].unique()
```

```python
# label type 
df_copy['type'] = df['type'].map({'white': 0, 'red': 1})
```

```python
df_copy.columns.tolist()
```

```python
df_copy = df_copy.drop(df_copy.columns[-1] , axis=1)
```

```python
df_copy.info()
```

```python
# preprocessing 

df_copy['target'] = np.where(df_copy['quality']>5, 1, 0)
df2 = df_copy.drop(['quality'],axis=1)
X = df2.drop(['target'],axis=1)
y = df2[['target']]
```

```python
# initializing random forest
rf = RandomForestClassifier()
```

```python
X
```

```python
y
```

```python
# grid search cv
grid_space={'max_depth':[3,5,10,None],
              'n_estimators':[10,100,200],
              'max_features':[1,3,5,7],
              'min_samples_leaf':[1,2,3],
              'min_samples_split':[1,2,3]
           }

grid = GridSearchCV(rf,param_grid=grid_space,cv=3,scoring='accuracy')
model_grid = grid.fit(X,y)
```

```python
# grid search results
print('Best grid search hyperparameters are: '+str(model_grid.best_params_))
print('Best grid search score is: '+str(model_grid.best_score_))
```

```python
# random search cv
rs_space={'max_depth':list(np.arange(10, 100, step=10)) + [None],
              'n_estimators':np.arange(10, 500, step=50),
              'max_features':randint(1,7),
              'criterion':['gini','entropy'],
              'min_samples_leaf':randint(1,4),
              'min_samples_split':np.arange(2, 10, step=2)
          }
```

```python
rf = RandomForestClassifier()

rf_random = RandomizedSearchCV(rf, rs_space, n_iter=500, scoring='accuracy', n_jobs=-1, cv=3)
model_random = rf_random.fit(X,y)
```

```python
# random random search results
print('Best random search hyperparameters are: '+str(model_random.best_params_))
print('Best random search score is: '+str(model_random.best_score_))
```
